```python
from pathlib import Path
import time

import pandas as pd
import yfinance as yf
```

```python
COMPANY_FILE = Path("/mnt/primary/sp500.csv")
EA_FILE = Path("/mnt/primary/Electronic Arts Stock Price History.csv")

OUTPUT_DIR = Path("/mnt/primary/Baseline/baseline data")
INDIVIDUAL_DIR = OUTPUT_DIR / "individual"
COMBINED_FILE = OUTPUT_DIR / "sp500_100_companies_10years.csv"
FAILED_FILE = OUTPUT_DIR / "failed_tickers.csv"
COVERAGE_FILE = OUTPUT_DIR / "coverage_summary.csv"

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
INDIVIDUAL_DIR.mkdir(parents=True, exist_ok=True)
```

```python
companies = pd.read_csv(COMPANY_FILE)

if "Symbol" not in companies.columns:
    raise ValueError("The company file must contain a 'Symbol' column.")

companies["Symbol"] = companies["Symbol"].astype(str).str.strip().str.upper()
companies = companies.drop_duplicates(subset=["Symbol"]).reset_index(drop=True)

print("Unique companies:", len(companies))
```

```python
def to_yahoo_symbol(symbol: str) -> str:
    return symbol.replace(".", "-")

companies["DownloadSymbol"] = companies["Symbol"].apply(to_yahoo_symbol)
```

```python
end_date = pd.to_datetime("2026-07-14")
start_date = pd.to_datetime("2016-07-14")

print("Start:", start_date.date())
print("End:", (end_date - pd.Timedelta(days=1)).date())
```

```python
def parse_volume(value):
    if pd.isna(value): return pd.NA
    value = str(value).strip().replace(",", "")
    if value in {"", "-", "nan"}: return pd.NA
    multipliers = {"K": 1_000, "M": 1_000_000, "B": 1_000_000_000}
    return float(value[:-1]) * multipliers[value[-1]] if value[-1] in multipliers else float(value)

def load_ea_data():
    if not EA_FILE.exists(): raise FileNotFoundError(f"EA fallback file not found: {EA_FILE}")
    data = pd.read_csv(EA_FILE).rename(columns={"Price": "Close", "Vol.": "Volume"})
    data["Date"] = pd.to_datetime(data["Date"], errors="coerce")
    data = data[(data["Date"] >= pd.Timestamp("2016-07-14")) & (data["Date"] <= pd.Timestamp("2026-07-14"))].copy()
    for column in ["Open", "High", "Low", "Close"]:
        data[column] = pd.to_numeric(data[column].astype(str).str.replace(",", "", regex=False), errors="coerce")
    data["Volume"] = data["Volume"].apply(parse_volume)
    data["Adj Close"] = data["Close"]
    return data

def prepare_data(data, original_symbol, download_symbol):
    if isinstance(data.columns, pd.MultiIndex): data.columns = data.columns.get_level_values(0)
    if "Date" not in data.columns: data = data.reset_index()
    if "Datetime" in data.columns: data = data.rename(columns={"Datetime": "Date"})
    data["Date"] = pd.to_datetime(data["Date"], errors="coerce")
    if data["Date"].dt.tz is not None: data["Date"] = data["Date"].dt.tz_localize(None)
    data["Symbol"] = original_symbol
    data["DownloadSymbol"] = download_symbol
    if "Adj Close" not in data.columns: data["Adj Close"] = data["Close"]
    required_columns = ["Date", "Symbol", "DownloadSymbol", "Open", "High", "Low", "Close", "Adj Close", "Volume"]
    return data[required_columns].dropna(subset=["Date", "Close"]).drop_duplicates(subset=["Symbol", "Date"], keep="last").sort_values("Date").reset_index(drop=True)

all_data = []
failed = []

for index, row in companies.iterrows():
    original_symbol, download_symbol = row["Symbol"], row["DownloadSymbol"]
    print(f"[{index + 1}/{len(companies)}] Downloading {original_symbol}")

    try:
        if original_symbol == "EA":
            print("  Using local EA historical CSV")
            data = load_ea_data()
        else:
            data = yf.download(download_symbol, start=start_date.strftime("%Y-%m-%d"), end=end_date.strftime("%Y-%m-%d"), interval="1d", auto_adjust=False, actions=False, progress=False, threads=False, timeout=30)

        if data is None or data.empty: raise ValueError("No data returned.")
        data = prepare_data(data, original_symbol, download_symbol)
        if data.empty: raise ValueError("No valid rows after cleaning.")

        output_path = INDIVIDUAL_DIR / f"{download_symbol}.csv"
        data.to_csv(output_path, index=False)
        all_data.append(data)
        print(f"  Success: {len(data)} rows, {data['Date'].min().date()} to {data['Date'].max().date()}")

    except Exception as error:
        failed.append({"Symbol": original_symbol, "DownloadSymbol": download_symbol, "Error": str(error)})
        print(f"  Failed: {error}")

    time.sleep(0.5)
```

```python
if not all_data:
    raise RuntimeError("No company data downloaded successfully.")

combined_data = pd.concat(all_data, ignore_index=True)
combined_data = combined_data.drop_duplicates(subset=["Symbol", "Date"], keep="last").sort_values(["Symbol", "Date"]).reset_index(drop=True)
combined_data.to_csv(COMBINED_FILE, index=False)
pd.DataFrame(failed, columns=["Symbol", "DownloadSymbol", "Error"]).to_csv(FAILED_FILE, index=False)

print("Companies downloaded:", combined_data["Symbol"].nunique())
print("Total rows:", len(combined_data))
print("Saved:", COMBINED_FILE)
```

```python
coverage = combined_data.groupby("Symbol").agg(Rows=("Date", "count"), FirstDate=("Date", "min"), LastDate=("Date", "max"), MissingOpen=("Open", lambda x: int(x.isna().sum())), MissingHigh=("High", lambda x: int(x.isna().sum())), MissingLow=("Low", lambda x: int(x.isna().sum())), MissingClose=("Close", lambda x: int(x.isna().sum())), MissingAdjustedClose=("Adj Close", lambda x: int(x.isna().sum())), MissingVolume=("Volume", lambda x: int(x.isna().sum()))).reset_index().sort_values("Rows")

coverage.to_csv(COVERAGE_FILE, index=False)
display(coverage)
```

```python
requested = set(companies["Symbol"])
downloaded = set(combined_data["Symbol"])
missing = sorted(requested - downloaded)

print("Requested companies:", len(requested))
print("Downloaded companies:", len(downloaded))
print("Missing companies:", missing)
```
