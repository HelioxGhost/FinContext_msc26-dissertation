```python
def safe_filename(value: str) -> str:
    value = str(value).strip()
    value = re.sub('[<>:"/\\\\|?*]', '_', value)
    value = re.sub('\\s+', ' ', value)
    return value.rstrip('. ')

def normalise_company_name(value: str) -> str:
    value = str(value).lower().strip()
    value = value.replace('&', 'and')
    value = re.sub('[^a-z0-9]+', '', value)
    return value

def serialise_json(data: Any) -> str:
    return json.dumps(data, indent=2, ensure_ascii=False, default=str)

def utc_timestamp_to_text(timestamp: Any) -> Optional[str]:
    try:
        timestamp = int(timestamp)
        if timestamp <= 0:
            return None
        return datetime.fromtimestamp(timestamp, tz=timezone.utc).isoformat()
    except (TypeError, ValueError, OSError):
        return None

def resolve_analysis_date(target_date: Optional[str]=None) -> date:
    if target_date is None:
        return datetime.now().date()
    return datetime.strptime(target_date, '%Y-%m-%d').date()

def percentage_change(current_value: Any, previous_value: Any) -> Optional[float]:
    """
    Calculate percentage change safely.
    """
    try:
        current_number = float(current_value)
        previous_number = float(previous_value)
        if previous_number == 0:
            return None
        return (current_number - previous_number) / abs(previous_number) * 100
    except (TypeError, ValueError):
        return None
```

```python
def load_companies(csv_path: Path) -> pd.DataFrame:
    if not csv_path.exists():
        raise FileNotFoundError(f'Company CSV was not found: {csv_path}')
    companies = pd.read_csv(csv_path)
    column_aliases = {'name': 'company_name', 'company': 'company_name', 'security': 'company_name', 'symbol': 'ticker', 'sector': 'category', 'gics sector': 'category', 'gics_sector': 'category'}
    rename_mapping = {}
    for column in companies.columns:
        normalised_column = str(column).strip().lower()
        rename_mapping[column] = column_aliases.get(normalised_column, normalised_column)
    companies = companies.rename(columns=rename_mapping)
    required_columns = {'company_name', 'ticker', 'category'}
    missing_columns = required_columns - set(companies.columns)
    if missing_columns:
        raise ValueError('The CSV is missing the following columns: ' + ', '.join(sorted(missing_columns)))
    companies = companies[['company_name', 'ticker', 'category']].copy()
    companies = companies.dropna(subset=['company_name', 'ticker'])
    companies['company_name'] = companies['company_name'].astype(str).str.strip()
    companies['ticker'] = companies['ticker'].astype(str).str.strip().str.upper()
    companies['category'] = companies['category'].fillna('Unknown').astype(str).str.strip()
    companies = companies.drop_duplicates(subset=['ticker'], keep='first')
    return companies.reset_index(drop=True)
```

```python
def build_question_file_index(questions_directory: Path) -> dict[str, Path]:
    if not questions_directory.exists():
        raise FileNotFoundError(f'Questions directory does not exist: {questions_directory}')
    question_files = list(questions_directory.glob('*.txt'))
    if not question_files:
        raise FileNotFoundError(f'No TXT question files were found in {questions_directory}')
    index = {}
    for path in question_files:
        index[normalise_company_name(path.stem)] = path
    return index

def find_question_file(company_name: str, ticker: str, question_file_index: dict[str, Path]) -> Optional[Path]:
    possible_names = [company_name, safe_filename(company_name), ticker, f'{company_name} {ticker}', f'{ticker} {company_name}']
    for possible_name in possible_names:
        key = normalise_company_name(possible_name)
        if key in question_file_index:
            return question_file_index[key]
    company_key = normalise_company_name(company_name)
    partial_matches = [path for key, path in question_file_index.items() if company_key in key or key in company_key]
    if len(partial_matches) == 1:
        return partial_matches[0]
    return None

def load_questions(question_file: Path) -> list[str]:
    text = question_file.read_text(encoding='utf-8', errors='replace').strip()
    if not text:
        raise ValueError(f'Question file is empty: {question_file}')
    numbered_pattern = re.compile('(?:^|\\n)\\s*(?:question\\s*)?(\\d{1,3})\\s*[\\.\\):\\-]\\s*(.*?)(?=(?:\\n\\s*(?:question\\s*)?\\d{1,3}\\s*[\\.\\):\\-]\\s*)|\\Z)', flags=re.IGNORECASE | re.DOTALL)
    matches = numbered_pattern.findall(text)
    if matches:
        questions = [re.sub('\\s+', ' ', question_text).strip() for _, question_text in matches if question_text.strip()]
    else:
        questions = [re.sub('\\s+', ' ', line).strip() for line in text.splitlines() if line.strip()]
    if not questions:
        raise ValueError(f'No questions could be extracted from {question_file}')
    return questions
```

```python
def finnhub_get(endpoint: str, params: Optional[dict[str, Any]]=None) -> Any:
    params = dict(params or {})
    params['token'] = FINNHUB_API_KEY
    url = f"{FINNHUB_BASE_URL}/{endpoint.lstrip('/')}"
    for attempt in range(1, MAX_FINNHUB_RETRIES + 1):
        try:
            response = finnhub_session.get(url, params=params, timeout=45)
            if response.status_code == 429:
                wait_time = min(60, 10 * attempt)
                print(f'        Finnhub rate limit reached. Waiting {wait_time} seconds...')
                time.sleep(wait_time)
                continue
            response.raise_for_status()
            data = response.json()
            time.sleep(FINNHUB_REQUEST_DELAY)
            return data
        except requests.RequestException as error:
            if attempt == MAX_FINNHUB_RETRIES:
                raise RuntimeError(f'Finnhub request failed for {endpoint}: {error}') from error
            wait_time = min(60, 2 ** attempt + random.uniform(0, 1))
            print(f'        Finnhub request failed. Retrying after {wait_time:.1f} seconds...')
            time.sleep(wait_time)
        except ValueError as error:
            raise RuntimeError(f'Finnhub returned invalid JSON for endpoint {endpoint}') from error
    raise RuntimeError(f'Finnhub request failed: {endpoint}')
```

```python
IMPORTANT_METRICS = ['10DayAverageTradingVolume', '13WeekPriceReturnDaily', '26WeekPriceReturnDaily', '3MonthAverageTradingVolume', '52WeekHigh', '52WeekHighDate', '52WeekLow', '52WeekLowDate', '52WeekPriceReturnDaily', '5DayPriceReturnDaily', 'assetTurnoverAnnual', 'beta', 'bookValuePerShareAnnual', 'cashFlowPerShareAnnual', 'currentRatioAnnual', 'dividendGrowthRate5Y', 'dividendPerShareAnnual', 'dividendYieldIndicatedAnnual', 'ebitdPerShareAnnual', 'epsAnnual', 'epsGrowth3Y', 'epsGrowth5Y', 'epsGrowthQuarterlyYoy', 'grossMarginAnnual', 'inventoryTurnoverAnnual', 'longTermDebtEquityAnnual', 'marketCapitalization', 'netDebtAnnual', 'netProfitMarginAnnual', 'operatingMarginAnnual', 'payoutRatioAnnual', 'pbAnnual', 'pcfShareAnnual', 'peAnnual', 'peBasicExclExtraTTM', 'peExclExtraAnnual', 'priceRelativeToS&P50013Week', 'priceRelativeToS&P50026Week', 'priceRelativeToS&P5004Week', 'priceRelativeToS&P50052Week', 'psAnnual', 'quickRatioAnnual', 'receivablesTurnoverAnnual', 'revenueGrowth3Y', 'revenueGrowth5Y', 'revenueGrowthQuarterlyYoy', 'revenuePerShareAnnual', 'roaRfy', 'roaTTM', 'roeRfy', 'roeTTM', 'roiAnnual', 'salesPerShareAnnual', 'totalDebtEquityAnnual']

def filter_basic_financials(data: Any) -> dict[str, Any]:
    if not isinstance(data, dict):
        return {}
    metrics = data.get('metric', {})
    selected_metrics = {metric_name: metrics.get(metric_name) for metric_name in IMPORTANT_METRICS if metric_name in metrics}
    filtered_series = {}
    series = data.get('series', {})
    if isinstance(series, dict):
        for period_name, period_values in series.items():
            if not isinstance(period_values, dict):
                continue
            filtered_series[period_name] = {}
            for metric_name, values in period_values.items():
                if isinstance(values, list):
                    filtered_series[period_name][metric_name] = values[-8:]
                else:
                    filtered_series[period_name][metric_name] = values
    return {'symbol': data.get('symbol'), 'metric_type': data.get('metricType'), 'metrics': selected_metrics, 'recent_series': filtered_series}
```

```python
def clean_news(news_items: Any, maximum_items: int=20) -> list[dict[str, Any]]:
    if not isinstance(news_items, list):
        return []
    cleaned = []
    for item in news_items[:maximum_items]:
        if not isinstance(item, dict):
            continue
        cleaned.append({'datetime_utc': utc_timestamp_to_text(item.get('datetime')), 'headline': item.get('headline'), 'summary': item.get('summary'), 'source': item.get('source'), 'category': item.get('category'), 'related': item.get('related'), 'url': item.get('url')})
    return cleaned

def clean_recommendations(data: Any) -> list[dict[str, Any]]:
    if not isinstance(data, list):
        return []
    return [{'period': row.get('period'), 'strong_buy': row.get('strongBuy'), 'buy': row.get('buy'), 'hold': row.get('hold'), 'sell': row.get('sell'), 'strong_sell': row.get('strongSell')} for row in data[:12] if isinstance(row, dict)]

def clean_earnings(data: Any) -> list[dict[str, Any]]:
    """
    Clean earnings surprise data.
    """
    if not isinstance(data, list):
        return []
    return [{'period': row.get('period'), 'quarter': row.get('quarter'), 'year': row.get('year'), 'actual_eps': row.get('actual'), 'estimated_eps': row.get('estimate'), 'surprise': row.get('surprise'), 'surprise_percent': row.get('surprisePercent'), 'symbol': row.get('symbol')} for row in data[:8] if isinstance(row, dict)]
```

```python
def snapshot_file_path(snapshot_date: str, company_name: str, ticker: str) -> Path:
    directory = SNAPSHOTS_DIRECTORY / snapshot_date
    filename = f'{safe_filename(ticker)}_snapshot.json'
    return directory / filename

def save_snapshot(snapshot: dict[str, Any]) -> Path:
    path = snapshot_file_path(snapshot_date=snapshot['snapshot_date'], company_name=snapshot['company_name'], ticker=snapshot['ticker'])
    path.parent.mkdir(parents=True, exist_ok=True)
    path.write_text(serialise_json(snapshot), encoding='utf-8')
    return path

def load_snapshot(snapshot_path: Path) -> dict[str, Any]:
    try:
        return json.loads(snapshot_path.read_text(encoding='utf-8'))
    except json.JSONDecodeError as error:
        raise ValueError(f'Invalid snapshot JSON: {snapshot_path}') from error
```

```python
def find_snapshot_file(snapshot_date: str, company_name: str, ticker: str) -> Optional[Path]:
    date_directory = SNAPSHOTS_DIRECTORY / snapshot_date
    if not date_directory.exists():
        return None
    safe_company = safe_filename(company_name)
    safe_ticker = safe_filename(ticker)
    exact_candidates = [date_directory / f'{safe_ticker}_snapshot.json', date_directory / f'{safe_company}_{safe_ticker}_snapshot.json', date_directory / f'{safe_company}_snapshot.json']
    for candidate in exact_candidates:
        if candidate.exists():
            return candidate
    target_ticker = str(ticker).strip().upper()
    target_company = normalise_company_name(company_name)
    metadata_matches = []
    for snapshot_path in date_directory.glob('*.json'):
        if 'snapshot' not in snapshot_path.stem.lower():
            continue
        try:
            snapshot_data = json.loads(snapshot_path.read_text(encoding='utf-8'))
        except (OSError, UnicodeDecodeError, json.JSONDecodeError):
            continue
        stored_ticker = str(snapshot_data.get('ticker', '')).strip().upper()
        stored_company = normalise_company_name(snapshot_data.get('company_name', ''))
        ticker_matches = stored_ticker == target_ticker
        company_matches = stored_company == target_company
        if ticker_matches or (not stored_ticker and company_matches):
            metadata_matches.append(snapshot_path)
    if len(metadata_matches) == 1:
        return metadata_matches[0]
    if len(metadata_matches) > 1:
        exact_company_matches = []
        for snapshot_path in metadata_matches:
            try:
                snapshot_data = json.loads(snapshot_path.read_text(encoding='utf-8'))
                stored_company = normalise_company_name(snapshot_data.get('company_name', ''))
                if stored_company == target_company:
                    exact_company_matches.append(snapshot_path)
            except Exception:
                continue
        if len(exact_company_matches) == 1:
            return exact_company_matches[0]
        print(f'Warning: multiple snapshots matched {company_name} ({ticker}) on {snapshot_date}:')
        for path in metadata_matches:
            print(f'    {path.name}')
        return None
    target_filename_keys = {normalise_company_name(f'{company_name}_{ticker}_snapshot'), normalise_company_name(f'{company_name}_snapshot'), normalise_company_name(f'{ticker}_snapshot')}
    filename_matches = []
    for snapshot_path in date_directory.glob('*.json'):
        file_key = normalise_company_name(snapshot_path.stem)
        if file_key in target_filename_keys:
            filename_matches.append(snapshot_path)
    if len(filename_matches) == 1:
        return filename_matches[0]
    if len(filename_matches) > 1:
        print(f'Warning: multiple filename matches found for {company_name} ({ticker}) on {snapshot_date}:')
        for path in filename_matches:
            print(f'    {path.name}')
    return None

def list_snapshot_dates() -> list[str]:
    if not SNAPSHOTS_DIRECTORY.exists():
        return []
    snapshot_dates = []
    for path in SNAPSHOTS_DIRECTORY.iterdir():
        if not path.is_dir():
            continue
        try:
            datetime.strptime(path.name, '%Y-%m-%d')
            snapshot_dates.append(path.name)
        except ValueError:
            continue
    return sorted(snapshot_dates)
```

```python
def fetch_finnhub_snapshot(company_name: str, ticker: str, category: str, analysis_date: date) -> dict[str, Any]:
    snapshot_date = analysis_date.isoformat()
    news_start = analysis_date - timedelta(days=NEWS_LOOKBACK_DAYS)
    snapshot = {'snapshot_date': snapshot_date, 'retrieved_at_utc': datetime.now(timezone.utc).isoformat(), 'company_name': company_name, 'ticker': ticker, 'category': category, 'data': {}, 'endpoint_errors': {}}
    endpoint_requests = {'quote': ('quote', {'symbol': ticker}), 'company_profile': ('stock/profile2', {'symbol': ticker}), 'basic_financials': ('stock/metric', {'symbol': ticker, 'metric': 'all'}), 'company_news': ('company-news', {'symbol': ticker, 'from': news_start.isoformat(), 'to': analysis_date.isoformat()}), 'recommendation_trends': ('stock/recommendation', {'symbol': ticker}), 'earnings_surprises': ('stock/earnings', {'symbol': ticker, 'limit': 8})}
    for data_name, request_data in endpoint_requests.items():
        endpoint, params = request_data
        try:
            print(f'        Fetching {data_name}...')
            result = finnhub_get(endpoint=endpoint, params=params)
            if data_name == 'basic_financials':
                result = filter_basic_financials(result)
            elif data_name == 'company_news':
                result = clean_news(result)
            elif data_name == 'recommendation_trends':
                result = clean_recommendations(result)
            elif data_name == 'earnings_surprises':
                result = clean_earnings(result)
            snapshot['data'][data_name] = result
        except Exception as error:
            print(f'        Warning: {data_name} failed: {error}')
            snapshot['endpoint_errors'][data_name] = str(error)
            snapshot['data'][data_name] = None
    quote = snapshot['data'].get('quote')
    if isinstance(quote, dict):
        quote_timestamp = utc_timestamp_to_text(quote.get('t'))
        snapshot['data']['quote_interpreted'] = {'current_price': quote.get('c'), 'daily_change': quote.get('d'), 'daily_percent_change': quote.get('dp'), 'daily_high': quote.get('h'), 'daily_low': quote.get('l'), 'daily_open': quote.get('o'), 'previous_close': quote.get('pc'), 'market_timestamp_utc': quote_timestamp}
    return snapshot
```

```python
def fetch_all_company_snapshots(target_date: Optional[str]=None) -> pd.DataFrame:
    analysis_date = resolve_analysis_date(target_date)
    snapshot_date = analysis_date.isoformat()
    print('=' * 80)
    print('PHASE 1: FETCHING ALL FINNHUB SNAPSHOTS')
    print(f'Snapshot date: {snapshot_date}')
    print('=' * 80)
    companies = load_companies(COMPANIES_CSV)
    if COMPANY_LIMIT is not None:
        companies = companies.head(COMPANY_LIMIT)
    results = []
    total_companies = len(companies)
    for row_index, row in companies.iterrows():
        company_number = row_index + 1
        company_name = row['company_name']
        ticker = row['ticker']
        category = row['category']
        print()
        print(f'[{company_number}/{total_companies}] {company_name} ({ticker})')
        existing_snapshot_path = find_snapshot_file(snapshot_date=snapshot_date, company_name=company_name, ticker=ticker)
        if existing_snapshot_path is not None and (not OVERWRITE_EXISTING_SNAPSHOTS):
            print('    Snapshot already exists. Skipping Finnhub calls.')
            results.append({'snapshot_date': snapshot_date, 'company_name': company_name, 'ticker': ticker, 'status': 'skipped_existing', 'snapshot_file': str(existing_snapshot_path), 'error': None})
            continue
        try:
            snapshot = fetch_finnhub_snapshot(company_name=company_name, ticker=ticker, category=category, analysis_date=analysis_date)
            saved_snapshot_path = save_snapshot(snapshot)
            print(f'    Snapshot saved to {saved_snapshot_path}')
            results.append({'snapshot_date': snapshot_date, 'company_name': company_name, 'ticker': ticker, 'status': 'completed', 'snapshot_file': str(saved_snapshot_path), 'error': None})
        except Exception as error:
            print(f'    ERROR: {error}')
            results.append({'snapshot_date': snapshot_date, 'company_name': company_name, 'ticker': ticker, 'status': 'failed', 'snapshot_file': None, 'error': str(error)})
    results_dataframe = pd.DataFrame(results)
    log_directory = SNAPSHOTS_DIRECTORY / snapshot_date
    log_directory.mkdir(parents=True, exist_ok=True)
    log_path = log_directory / f'snapshot_processing_log_{snapshot_date}.csv'
    results_dataframe.to_csv(log_path, index=False)
    print()
    print('=' * 80)
    print('SNAPSHOT PHASE FINISHED')
    print(f'Snapshot log: {log_path}')
    if not results_dataframe.empty:
        print()
        print(results_dataframe['status'].value_counts())
    return results_dataframe
```

```python
def find_previous_snapshot_from_storage(company_name: str, ticker: str, current_snapshot_date: str) -> Optional[dict[str, Any]]:
    current_date = datetime.strptime(current_snapshot_date, '%Y-%m-%d').date()
    candidate_snapshots = []
    for snapshot_date in list_snapshot_dates():
        candidate_date = datetime.strptime(snapshot_date, '%Y-%m-%d').date()
        if candidate_date >= current_date:
            continue
        snapshot_path = find_snapshot_file(snapshot_date=snapshot_date, company_name=company_name, ticker=ticker)
        if snapshot_path is not None:
            candidate_snapshots.append((candidate_date, snapshot_path))
    if not candidate_snapshots:
        return None
    _, latest_path = max(candidate_snapshots, key=lambda item: item[0])
    return load_snapshot(latest_path)
```

```python
def compare_snapshots(current_snapshot: dict[str, Any], previous_snapshot: Optional[dict[str, Any]]) -> dict[str, Any]:
    if previous_snapshot is None:
        return {'comparison_available': False, 'message': "No earlier local snapshot was found. Use the current quote's daily change and daily percentage change fields."}
    current_data = current_snapshot.get('data', {})
    previous_data = previous_snapshot.get('data', {})
    current_quote = current_data.get('quote_interpreted') or {}
    previous_quote = previous_data.get('quote_interpreted') or {}
    current_metrics = (current_data.get('basic_financials') or {}).get('metrics', {})
    previous_metrics = (previous_data.get('basic_financials') or {}).get('metrics', {})
    metric_changes = {}
    all_metric_names = sorted(set(current_metrics) | set(previous_metrics))
    for metric_name in all_metric_names:
        current_value = current_metrics.get(metric_name)
        previous_value = previous_metrics.get(metric_name)
        if current_value != previous_value:
            metric_changes[metric_name] = {'previous': previous_value, 'current': current_value, 'percentage_change': percentage_change(current_value, previous_value)}
    current_news = current_data.get('company_news') or []
    previous_news = previous_data.get('company_news') or []
    previous_news_keys = {(item.get('headline'), item.get('datetime_utc')) for item in previous_news if isinstance(item, dict)}
    new_news = [item for item in current_news if isinstance(item, dict) and (item.get('headline'), item.get('datetime_utc')) not in previous_news_keys]
    return {'comparison_available': True, 'previous_snapshot_date': previous_snapshot.get('snapshot_date'), 'current_snapshot_date': current_snapshot.get('snapshot_date'), 'price_comparison': {'previous_snapshot_current_price': previous_quote.get('current_price'), 'current_snapshot_current_price': current_quote.get('current_price'), 'percentage_change_between_snapshots': percentage_change(current_quote.get('current_price'), previous_quote.get('current_price')), 'current_market_daily_change': current_quote.get('daily_change'), 'current_market_daily_percent_change': current_quote.get('daily_percent_change')}, 'changed_financial_metrics': metric_changes, 'new_company_news_since_previous_snapshot': new_news}
```

```python
def split_into_batches(items: list[Any], batch_size: int) -> list[list[Any]]:
    return [items[index:index + batch_size] for index in range(0, len(items), batch_size)]

def format_questions_for_prompt(questions: list[str], starting_question_number: int) -> str:
    formatted_questions = []
    for offset, question in enumerate(questions):
        question_number = starting_question_number + offset
        formatted_questions.append(f'{question_number}. {question}')
    return '\n\n'.join(formatted_questions)
```

```python
def build_company_prompt(company_name: str, ticker: str, category: str, questions: list[str], starting_question_number: int, current_snapshot: dict[str, Any], daily_comparison: dict[str, Any]) -> str:
    ending_question_number = starting_question_number + len(questions) - 1
    questions_text = format_questions_for_prompt(questions=questions, starting_question_number=starting_question_number)
    return f"""\nYou are a careful financial research assistant participating in a longitudinal experiment that studies whether daily company information may help predict future stock performance.\n\nCOMPANY\n\nCompany name: {company_name}\nTicker: {ticker}\nCategory: {category}\nAnalysis date: {current_snapshot['snapshot_date']}\n\nTASK\n\nAnswer every supplied question as usefully and analytically as possible.\n\nThis request contains question numbers {starting_question_number} to {ending_question_number}.\n\nYou must return one answer section for every supplied question number.\n\nNever skip a question number, even when the evidence is limited.\n\nEVIDENCE HIERARCHY\n\nUse evidence in this order:\n\n1. Current Finnhub snapshot.\n2. Comparison with the previous stored snapshot.\n3. Recent Finnhub company news.\n4. Historical Finnhub financial metrics and earnings data.\n5. Facts and figures explicitly included in the question.\n6. Stable general knowledge about the company and its business model.\n7. Clearly labelled financial reasoning and inference.\n\nIMPORTANT RULES\n\n- Treat current Finnhub data as the primary evidence for current claims.\n- Facts explicitly stated in the question may be treated as supplied context.\n- Do not invent current values, dates, transactions, events or legal outcomes.\n- Do not present an inference as a confirmed fact.\n- A missing value does not mean that the value is zero.\n- Do not provide personalised financial advice.\n- Do not provide an unsupported buy, sell or hold recommendation.\n\nWHEN EXACT DATA IS UNAVAILABLE\n\nDo not answer only with:\n\n"The supplied Finnhub data does not provide enough information."\n\nInstead:\n\n1. Briefly state which exact figure is unavailable.\n2. Use the facts included in the question as context.\n3. Use any related Finnhub metrics, news or trends.\n4. Provide a directional or scenario-based assessment.\n5. Clearly label the assessment as an inference.\n6. Explain what additional evidence would be needed for a definitive answer.\n\nFor projected or hypothetical questions, provide a reasonable scenario-based answer rather than refusing to answer.\n\nANALYTICAL APPROACH\n\nFor each question:\n\n1. Identify the financial or strategic issue.\n2. Extract relevant facts from the question.\n3. Find related evidence in the current snapshot.\n4. Check for changes from the previous snapshot.\n5. Use stable company knowledge where helpful.\n6. Determine whether the implication is positive, negative, mixed or neutral.\n7. Explain the possible short-, medium- or long-term stock relevance.\n8. State the confidence level.\n\nOUTPUT FORMAT\n\nReturn exactly one numbered answer block for every question in this request.\n\nDo not reproduce the full question.\n\nUse the original question number.\n\nUse this exact structure:\n\nANSWER {starting_question_number}\n\nDIRECT ANSWER\n<direct answer in approximately one to three paragraphs>\n\nAVAILABLE EVIDENCE\n- <relevant Finnhub evidence>\n- <relevant facts contained in the question>\n- <relevant daily change or recent news>\n\nANALYSIS AND INFERENCE\n<clearly distinguish verified evidence from inference>\n\nSTOCK-PERFORMANCE RELEVANCE\n<state whether the implication is positive, negative, mixed or neutral and explain the likely time horizon>\n\nCONFIDENCE\n<High, Medium or Low, followed by one sentence explaining why>\n\nThen continue with:\n\nANSWER {starting_question_number + 1}\n\nContinue until ANSWER {ending_question_number}.\n\nDo not add an introduction.\n\nDo not add a conclusion.\n\nQUESTIONS\n\n{questions_text}\n\nCURRENT FINNHUB SNAPSHOT\n\n{serialise_json(current_snapshot)}\n\nCOMPARISON WITH THE PREVIOUS STORED SNAPSHOT\n\n{serialise_json(daily_comparison)}\n""".strip()
```

```python
def answer_questions_with_llm(prompt: str) -> str:
    for attempt in range(1, MAX_LLM_RETRIES + 1):
        try:
            result = llm_client.responses.create(model=LLM_MODEL, input=prompt)
            output_text = result.output_text
            if not output_text or not output_text.strip():
                raise RuntimeError('The LLM returned an empty response.')
            return output_text.strip()
        except Exception as error:
            if attempt == MAX_LLM_RETRIES:
                raise RuntimeError(f'LLM request failed after {MAX_LLM_RETRIES} attempts: {error}') from error
            wait_time = min(60, 2 ** attempt * 5 + random.uniform(0, 2))
            print(f'        LLM request failed. Retrying after {wait_time:.1f} seconds...')
            time.sleep(wait_time)
    raise RuntimeError('The LLM request failed.')
```

```python
def extract_answer_blocks(output_text: str) -> dict[int, str]:
    blocks = {}
    pattern = re.compile('(?:^|\\n)\\s*(?:\\*{0,2})ANSWER\\s+(\\d+)(?:\\*{0,2})\\s*[:.\\-]?\\s*(.*?)(?=(?:\\n\\s*(?:\\*{0,2})ANSWER\\s+\\d+(?:\\*{0,2})\\s*[:.\\-]?)|\\Z)', flags=re.IGNORECASE | re.DOTALL)
    for match in pattern.finditer(output_text):
        question_number = int(match.group(1))
        answer_block = match.group(2).strip()
        if answer_block:
            blocks[question_number] = answer_block
    return blocks
```

```python
def answer_all_question_batches(company_name: str, ticker: str, category: str, questions: list[str], current_snapshot: dict[str, Any], daily_comparison: dict[str, Any]) -> tuple[dict[int, str], list[int]]:
    """
    Send questions to the LLM in smaller batches.
    """
    all_answer_blocks = {}
    question_batches = split_into_batches(questions, QUESTION_BATCH_SIZE)
    total_batches = len(question_batches)
    question_offset = 0
    for batch_index, batch in enumerate(question_batches, start=1):
        starting_question_number = question_offset + 1
        ending_question_number = starting_question_number + len(batch) - 1
        print(f'        Batch {batch_index}/{total_batches}: questions {starting_question_number}-{ending_question_number}')
        prompt = build_company_prompt(company_name=company_name, ticker=ticker, category=category, questions=batch, starting_question_number=starting_question_number, current_snapshot=current_snapshot, daily_comparison=daily_comparison)
        print(f'        Prompt length: {len(prompt):,} characters')
        raw_output = answer_questions_with_llm(prompt)
        batch_blocks = extract_answer_blocks(raw_output)
        expected_numbers = range(starting_question_number, ending_question_number + 1)
        for question_number in expected_numbers:
            if question_number in batch_blocks:
                all_answer_blocks[question_number] = batch_blocks[question_number]
        question_offset += len(batch)
    missing_answer_numbers = [question_number for question_number in range(1, len(questions) + 1) if question_number not in all_answer_blocks]
    return (all_answer_blocks, missing_answer_numbers)
```

```python
def build_final_answer_text(questions: list[str], answer_blocks: dict[int, str]) -> tuple[str, list[int]]:
    output_sections = []
    missing_answers = []
    for question_number, question in enumerate(questions, start=1):
        answer_block = answer_blocks.get(question_number)
        if not answer_block:
            missing_answers.append(question_number)
            answer_block = 'No answer was returned by the model for this question.'
        section = f'QUESTION {question_number}\n{question}\n\nANSWER\n{answer_block}'
        output_sections.append(section)
    separator = '\n\n' + '-' * 80 + '\n\n'
    final_text = separator.join(output_sections)
    return (final_text, missing_answers)
```

```python
def answer_file_path(company_name: str, answer_date: str) -> Path:
    """
    Return the answer file path.
    """
    date_directory = ANSWERS_DIRECTORY / answer_date
    filename = f'{safe_filename(company_name)}_answer_{answer_date}.txt'
    return date_directory / filename

def save_answer_file(company_name: str, ticker: str, category: str, answer_date: str, question_file: Path, answer_text: str, missing_answer_numbers: list[int], snapshot_path: Path) -> Path:
    """
    Save the final answer text.
    """
    path = answer_file_path(company_name=company_name, answer_date=answer_date)
    path.parent.mkdir(parents=True, exist_ok=True)
    if missing_answer_numbers:
        validation_status = 'WARNING - no answer block detected for question numbers: ' + ', '.join(map(str, missing_answer_numbers))
    else:
        validation_status = 'PASS - an answer block was detected for every question'
    file_content = f"COMPANY: {company_name}\nTICKER: {ticker}\nCATEGORY: {category}\nANALYSIS DATE: {answer_date}\nQUESTION FILE: {question_file}\nSNAPSHOT FILE: {snapshot_path}\nMODEL: {LLM_MODEL}\nVALIDATION: {validation_status}\n\n{'=' * 80}\n\n{answer_text}\n"
    path.write_text(file_content, encoding='utf-8')
    return path
```

```python
def generate_answer_from_stored_snapshot(company_name: str, ticker: str, category: str, snapshot_date: str, snapshot_path: Path, question_file_index: dict[str, Path]) -> dict[str, Any]:
    """
    Generate one company's answer using a stored snapshot.
    """
    output_path = answer_file_path(company_name=company_name, answer_date=snapshot_date)
    if output_path.exists() and (not OVERWRITE_EXISTING_ANSWERS):
        return {'snapshot_date': snapshot_date, 'company_name': company_name, 'ticker': ticker, 'status': 'skipped_existing_answer', 'snapshot_file': str(snapshot_path), 'answer_file': str(output_path), 'question_count': None, 'missing_answers': None, 'error': None}
    question_file = find_question_file(company_name=company_name, ticker=ticker, question_file_index=question_file_index)
    if question_file is None:
        raise FileNotFoundError(f'No question file was found for {company_name} ({ticker}).')
    questions = load_questions(question_file)
    print(f'    Loaded {len(questions)} questions from {question_file.name}')
    if len(questions) != 30:
        print(f'    Warning: expected 30 questions but found {len(questions)}.')
    current_snapshot = load_snapshot(snapshot_path)
    previous_snapshot = find_previous_snapshot_from_storage(company_name=company_name, ticker=ticker, current_snapshot_date=snapshot_date)
    if previous_snapshot:
        print(f"    Previous snapshot: {previous_snapshot.get('snapshot_date')}")
    else:
        print('    No previous snapshot available.')
    daily_comparison = compare_snapshots(current_snapshot=current_snapshot, previous_snapshot=previous_snapshot)
    print('    Sending question batches to the university LLM...')
    answer_blocks, initial_missing = answer_all_question_batches(company_name=company_name, ticker=ticker, category=category, questions=questions, current_snapshot=current_snapshot, daily_comparison=daily_comparison)
    final_answer_text, missing_answers = build_final_answer_text(questions=questions, answer_blocks=answer_blocks)
    if missing_answers:
        print('    Warning: no answer block was detected for question numbers: ' + ', '.join(map(str, missing_answers)))
    else:
        print('    Validation passed: all questions have an answer.')
    saved_answer_path = save_answer_file(company_name=company_name, ticker=ticker, category=category, answer_date=snapshot_date, question_file=question_file, answer_text=final_answer_text, missing_answer_numbers=missing_answers, snapshot_path=snapshot_path)
    print(f'    Answer saved to {saved_answer_path}')
    return {'snapshot_date': snapshot_date, 'company_name': company_name, 'ticker': ticker, 'status': 'completed', 'snapshot_file': str(snapshot_path), 'answer_file': str(saved_answer_path), 'question_count': len(questions), 'missing_answers': ','.join(map(str, missing_answers)) if missing_answers else None, 'error': None}
```

```python
def generate_all_missing_answers(target_date: Optional[str]=None, process_all_dates: bool=True) -> pd.DataFrame:
    print()
    print('=' * 80)
    print('PHASE 2: GENERATING ANSWERS FROM STORED SNAPSHOTS')
    print('=' * 80)
    companies = load_companies(COMPANIES_CSV)
    if COMPANY_LIMIT is not None:
        companies = companies.head(COMPANY_LIMIT)
    question_file_index = build_question_file_index(QUESTIONS_DIRECTORY)
    if process_all_dates:
        snapshot_dates = list_snapshot_dates()
    else:
        analysis_date = resolve_analysis_date(target_date)
        snapshot_dates = [analysis_date.isoformat()]
    if not snapshot_dates:
        print('No stored snapshot dates were found.')
        return pd.DataFrame()
    print('Snapshot dates to process:')
    for snapshot_date in snapshot_dates:
        print(f'    {snapshot_date}')
    all_results = []
    for snapshot_date in snapshot_dates:
        print()
        print('#' * 80)
        print(f'PROCESSING SNAPSHOT DATE: {snapshot_date}')
        print('#' * 80)
        date_results = []
        total_companies = len(companies)
        for row_index, row in companies.iterrows():
            company_number = row_index + 1
            company_name = row['company_name']
            ticker = row['ticker']
            category = row['category']
            print()
            print(f'[{company_number}/{total_companies}] {company_name} ({ticker})')
            existing_answer_path = answer_file_path(company_name=company_name, answer_date=snapshot_date)
            if existing_answer_path.exists() and (not OVERWRITE_EXISTING_ANSWERS):
                print('    Answer already exists. Skipping.')
                result = {'snapshot_date': snapshot_date, 'company_name': company_name, 'ticker': ticker, 'status': 'skipped_existing_answer', 'snapshot_file': None, 'answer_file': str(existing_answer_path), 'question_count': None, 'missing_answers': None, 'error': None}
                date_results.append(result)
                all_results.append(result)
                continue
            snapshot_path = find_snapshot_file(snapshot_date=snapshot_date, company_name=company_name, ticker=ticker)
            if snapshot_path is None:
                print('    No stored snapshot found. Skipping.')
                result = {'snapshot_date': snapshot_date, 'company_name': company_name, 'ticker': ticker, 'status': 'missing_snapshot', 'snapshot_file': None, 'answer_file': None, 'question_count': None, 'missing_answers': None, 'error': f'No stored snapshot found for {snapshot_date}'}
                date_results.append(result)
                all_results.append(result)
                continue
            try:
                result = generate_answer_from_stored_snapshot(company_name=company_name, ticker=ticker, category=category, snapshot_date=snapshot_date, snapshot_path=snapshot_path, question_file_index=question_file_index)
            except Exception as error:
                print(f'    ERROR: {error}')
                result = {'snapshot_date': snapshot_date, 'company_name': company_name, 'ticker': ticker, 'status': 'failed', 'snapshot_file': str(snapshot_path), 'answer_file': None, 'question_count': None, 'missing_answers': None, 'error': str(error)}
            date_results.append(result)
            all_results.append(result)
        date_results_dataframe = pd.DataFrame(date_results)
        answer_date_directory = ANSWERS_DIRECTORY / snapshot_date
        answer_date_directory.mkdir(parents=True, exist_ok=True)
        date_log_path = answer_date_directory / f'answer_processing_log_{snapshot_date}.csv'
        date_results_dataframe.to_csv(date_log_path, index=False)
        print()
        print(f'Date log saved to {date_log_path}')
    all_results_dataframe = pd.DataFrame(all_results)
    print()
    print('=' * 80)
    print('ANSWER PHASE FINISHED')
    if not all_results_dataframe.empty:
        print()
        print(all_results_dataframe['status'].value_counts())
    return all_results_dataframe
```

```python
def run_daily_pipeline(target_date: Optional[str]=None, recover_previous_missing_answers: bool=True) -> tuple[pd.DataFrame, pd.DataFrame]:
    """
    Run the complete pipeline.

    Phase 1:
    Fetch all company snapshots.

    Phase 2:
    Generate missing answer files using stored snapshots.
    """
    print('=' * 80)
    print('STARTING TWO-PHASE FINNHUB AND LLM PIPELINE')
    print('=' * 80)
    snapshot_results = fetch_all_company_snapshots(target_date=target_date)
    print()
    print('All Finnhub snapshot requests have finished.')
    print('Beginning LLM answer generation.')
    answer_results = generate_all_missing_answers(target_date=target_date, process_all_dates=recover_previous_missing_answers)
    return (snapshot_results, answer_results)
```

```python
snapshot_results, answer_results = run_daily_pipeline(target_date=None, recover_previous_missing_answers=True)
```
