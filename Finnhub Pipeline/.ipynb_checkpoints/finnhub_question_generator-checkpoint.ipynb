````python
import os
import re
import json
import time
import html
from pathlib import Path
from typing import Optional
import pandas as pd
from bs4 import BeautifulSoup
from openai import OpenAI
from tqdm.auto import tqdm
````

````python
CSV_PATH = Path('/mnt/primary/sp500.csv')
FILINGS_DIR = Path('/mnt/primary/10k-filings')
QUESTIONS_DIR = Path('mnt/primary/Finnhub Pipeline/finnhub_questions')
QUESTIONS_DIR.mkdir(parents=True, exist_ok=True)
BASE_URL = 'http://api.llm.apps.os.dcs.gla.ac.uk/v1'
API_KEY = os.environ['IDA_LLM_API_KEY']
MODEL_NAME = 'gemma-4-26b-a4b'
client = OpenAI(base_url=BASE_URL, api_key=API_KEY)
NUMBER_OF_QUESTIONS = 30
MAX_FILING_CHARACTERS = 100000
CHUNK_SIZE = 4000
DELAY_BETWEEN_REQUESTS = 2
MAX_RETRIES = 3
OVERWRITE_EXISTING = False
````

````python
companies = pd.read_csv(CSV_PATH)
required_columns = ['Symbol', 'Security', 'GICS Sector', 'GICS Sub-Industry']
missing_columns = [column for column in required_columns if column not in companies.columns]
if missing_columns:
    raise ValueError(f'The CSV is missing these required columns: {missing_columns}')
companies = companies.dropna(subset=['Symbol', 'Security']).copy()
companies['Symbol'] = companies['Symbol'].astype(str).str.strip()
companies['Security'] = companies['Security'].astype(str).str.strip()
print(f'Number of companies loaded: {len(companies)}')
display(companies.head())
````

````python
def normalize_ticker(ticker: str) -> str:
    return re.sub('[^A-Z0-9]', '', str(ticker).upper())

def extract_ticker_from_filename(file_path: Path) -> Optional[str]:
    match = re.match('^(.+?)_10-?K_', file_path.stem, flags=re.IGNORECASE)
    if match:
        return match.group(1)
    return None

def build_filing_index(filings_directory: Path) -> dict:
    if not filings_directory.exists():
        raise FileNotFoundError(f'10-K directory does not exist: {filings_directory.resolve()}')
    filing_index = {}
    accepted_extensions = {'.html', '.htm', '.txt'}
    for file_path in filings_directory.rglob('*'):
        if not file_path.is_file():
            continue
        if file_path.suffix.lower() not in accepted_extensions:
            continue
        ticker_from_filename = extract_ticker_from_filename(file_path)
        if ticker_from_filename is None:
            continue
        normalised_ticker = normalize_ticker(ticker_from_filename)
        filing_index.setdefault(normalised_ticker, []).append(file_path)
    return filing_index
````

````python
def extract_filing_date(file_path: Path) -> str:
    match = re.search('(\\d{4}-\\d{2}-\\d{2})', file_path.stem)
    return match.group(1) if match else ''

def find_company_filing(ticker: str, filing_index: dict) -> Optional[Path]:
    normalised_ticker = normalize_ticker(ticker)
    matches = filing_index.get(normalised_ticker, [])
    if not matches:
        return None
    matches = sorted(matches, key=extract_filing_date, reverse=True)
    return matches[0]
````

````python
filing_index = build_filing_index(FILINGS_DIR)
print(f'Ticker entries found in filing directory: {len(filing_index)}')
matched_filings = []
missing_filings = []
for _, row in companies.iterrows():
    ticker = row['Symbol']
    filing_path = find_company_filing(ticker, filing_index)
    if filing_path:
        matched_filings.append({'Ticker': ticker, 'Company': row['Security'], 'Filing': filing_path.name})
    else:
        missing_filings.append({'Ticker': ticker, 'Company': row['Security']})
print(f'Companies with filings: {len(matched_filings)}')
print(f'Companies without filings: {len(missing_filings)}')
if missing_filings:
    display(pd.DataFrame(missing_filings))
````

````python
def read_file_with_fallback_encoding(file_path: Path) -> str:
    encodings = ['utf-8', 'utf-8-sig', 'latin-1', 'cp1252']
    for encoding in encodings:
        try:
            return file_path.read_text(encoding=encoding, errors='strict')
        except UnicodeDecodeError:
            continue
    return file_path.read_text(encoding='utf-8', errors='ignore')
````

````python
def clean_filing_html(file_path: Path) -> str:
    raw_content = read_file_with_fallback_encoding(file_path)
    if file_path.suffix.lower() in {'.html', '.htm'}:
        soup = BeautifulSoup(raw_content, 'lxml')
        for element in soup(['script', 'style', 'noscript', 'svg', 'meta', 'link']):
            element.decompose()
        text = soup.get_text(separator=' ')
    else:
        text = raw_content
    text = html.unescape(text)
    text = text.replace('\xa0', ' ')
    text = re.sub('\\s+', ' ', text)
    return text.strip()
````

````python
IMPORTANT_FILING_TERMS = {'risk factors': 8, 'management discussion': 8, "management's discussion": 8, 'results of operations': 7, 'business': 4, 'revenue': 5, 'net income': 5, 'operating income': 5, 'operating margin': 5, 'gross margin': 5, 'cash flow': 5, 'free cash flow': 5, 'liquidity': 6, 'capital resources': 6, 'capital expenditure': 5, 'debt': 5, 'borrowings': 4, 'interest rate': 4, 'competition': 6, 'competitors': 5, 'market risk': 6, 'foreign currency': 4, 'supply chain': 5, 'customer concentration': 6, 'supplier concentration': 5, 'regulation': 5, 'regulatory': 5, 'litigation': 5, 'cybersecurity': 5, 'acquisition': 5, 'merger': 5, 'impairment': 4, 'segment': 4, 'geographic': 3, 'research and development': 4, 'seasonality': 4, 'outlook': 6, 'guidance': 6}
````

````python
def split_text_into_chunks(text: str, chunk_size: int=CHUNK_SIZE) -> list[str]:
    sentences = re.split('(?<=[.!?])\\s+', text)
    chunks = []
    current_chunk = []
    current_length = 0
    for sentence in sentences:
        sentence_length = len(sentence)
        if current_chunk and current_length + sentence_length > chunk_size:
            chunks.append(' '.join(current_chunk))
            current_chunk = []
            current_length = 0
        current_chunk.append(sentence)
        current_length += sentence_length + 1
    if current_chunk:
        chunks.append(' '.join(current_chunk))
    return chunks
````

````python
def score_filing_chunk(chunk: str) -> float:
    lower_chunk = chunk.lower()
    score = 0.0
    for term, weight in IMPORTANT_FILING_TERMS.items():
        occurrences = lower_chunk.count(term)
        score += min(occurrences, 4) * weight
    number_count = len(re.findall('\\$[\\d,.]+|\\d+(?:\\.\\d+)?%', chunk))
    score += min(number_count, 10) * 0.5
    return score
````

````python
def select_relevant_filing_text(filing_text: str, max_characters: int=MAX_FILING_CHARACTERS) -> str:
    if len(filing_text) <= max_characters:
        return filing_text
    chunks = split_text_into_chunks(filing_text)
    if not chunks:
        return filing_text[:max_characters]
    selected_chunks = []
    selected_indices = set()
    total_characters = 0
    initial_chunk_count = min(3, len(chunks))
    for index in range(initial_chunk_count):
        selected_chunks.append((index, chunks[index]))
        selected_indices.add(index)
        total_characters += len(chunks[index])
    scored_chunks = [(score_filing_chunk(chunk), index, chunk) for index, chunk in enumerate(chunks) if index not in selected_indices]
    scored_chunks.sort(key=lambda item: item[0], reverse=True)
    for score, index, chunk in scored_chunks:
        if total_characters + len(chunk) > max_characters:
            continue
        selected_chunks.append((index, chunk))
        selected_indices.add(index)
        total_characters += len(chunk)
        if total_characters >= max_characters:
            break
    selected_chunks.sort(key=lambda item: item[0])
    return '\n\n'.join((chunk for _, chunk in selected_chunks))
````

````python
def create_question_generation_prompt(company_name: str, ticker: str, sector: str, sub_industry: str, filing_date: str, filing_text: str) -> str:
    return f"""\nYou are a financial research question-generation system.\n\nYour task is to create exactly {NUMBER_OF_QUESTIONS} analytical questions\nfor monitoring and predicting the future stock performance of one publicly\nlisted company.\n\nCOMPANY INFORMATION\nCompany name: {company_name}\nTicker: {ticker}\nGICS sector: {sector}\nGICS sub-industry: {sub_industry}\n10-K filing date: {filing_date}\n\nPURPOSE\nThe questions will later be asked repeatedly using current financial,\nmarket, company, industry and macroeconomic information. Changes in their\nanswers will be studied to determine whether they contain signals about\nthe company's future stock performance.\n\nThe attached 10-K is supplied only as background for generating informed\nand company-specific questions. Do not answer the questions. Do not merely\nsummarise the filing.\n\nINSTRUCTIONS\n\n1. Produce exactly {NUMBER_OF_QUESTIONS} questions.\n\n2. Every item must be written as a complete question.\n\n3. Create questions specifically for {company_name}. Use:\n   - the company's name and ticker;\n   - its sector and sub-industry;\n   - material business characteristics visible in the 10-K;\n   - products, services, markets and business segments identified in the filing;\n   - revenue drivers and cost drivers;\n   - company-specific risks and opportunities;\n   - your existing general knowledge of the company where useful.\n\n4. The questions must be useful for evaluating or predicting stock\n   performance over future short-, medium- or long-term periods.\n\n5. Prefer questions whose answers could change over time and could therefore\n   be monitored repeatedly. Questions should examine measurable developments,\n   trends, events, comparisons or changes rather than static facts.\n\n6. Include a balanced collection of questions covering relevant areas such as:\n   - revenue growth and demand;\n   - pricing power and unit economics;\n   - margins, expenses and profitability;\n   - cash flow and capital expenditure;\n   - balance-sheet strength, debt and liquidity;\n   - business segments, products and geographic markets;\n   - customer, supplier or partner concentration;\n   - competition and market share;\n   - management guidance and earnings expectations;\n   - valuation and investor expectations;\n   - regulatory, legal, geopolitical and cybersecurity risks;\n   - technological disruption and innovation;\n   - acquisitions, divestitures and capital allocation;\n   - sector-specific economic indicators;\n   - macroeconomic sensitivity;\n   - potential positive and negative catalysts.\n\n7. Sector relevance is essential. Do not force irrelevant metrics onto the\n   company. For example, use subscriber metrics only when subscriptions are\n   material, same-store sales only for relevant retailers, and production\n   volumes only for companies where production is a meaningful driver.\n\n8. Make each question sufficiently precise that it could later be answered\n   using company reports, market data, news, analyst information or other\n   current sources.\n\n9. Questions should encourage analysis rather than simple fact retrieval.\n   Where appropriate, ask about:\n   - direction and magnitude of change;\n   - comparisons with prior periods;\n   - comparisons with management guidance;\n   - comparisons with competitors or sector benchmarks;\n   - possible effect on revenue, earnings, cash flow, valuation or investor\n     sentiment.\n\n10. Avoid:\n    - duplicate or near-duplicate questions;\n    - generic questions that could be copied unchanged to almost any company;\n    - questions asking only for the current stock price;\n    - questions asking for a direct stock-price prediction without supporting\n      analysis;\n    - questions based on unsupported assumptions;\n    - questions whose answers are permanently fixed;\n    - questions that reveal their answers inside the question;\n    - multi-part questions containing several unrelated topics.\n\n11. Do not cite page numbers because HTML filing pagination may be unreliable.\n\n12. Do not provide explanations, headings, categories, introductory text,\n    conclusions or answers.\n\nOUTPUT FORMAT\n\nReturn only a valid JSON array containing exactly {NUMBER_OF_QUESTIONS}\nstrings.\n\nRequired structure:\n\n[\n  "First complete question?",\n  "Second complete question?",\n  "Third complete question?"\n]\n\nDo not wrap the JSON in Markdown code fences.\n\n10-K FILING EXTRACT\n-------------------\n{filing_text}\n-------------------\nEND OF 10-K FILING EXTRACT\n""".strip()
````

````python
def remove_markdown_code_fence(text: str) -> str:
    text = text.strip()
    text = re.sub('^```(?:json)?\\s*', '', text, flags=re.IGNORECASE)
    text = re.sub('\\s*```$', '', text)
    return text.strip()
````

````python
def extract_json_array(response_text: str) -> list:
    cleaned_text = remove_markdown_code_fence(response_text)
    try:
        parsed = json.loads(cleaned_text)
    except json.JSONDecodeError:
        start_position = cleaned_text.find('[')
        end_position = cleaned_text.rfind(']')
        if start_position == -1 or end_position == -1:
            raise ValueError('No JSON array was found in the model response.')
        possible_json = cleaned_text[start_position:end_position + 1]
        parsed = json.loads(possible_json)
    if not isinstance(parsed, list):
        raise ValueError('The model response is not a JSON array.')
    return parsed
````

````python
def validate_questions(questions: list, expected_count: int=NUMBER_OF_QUESTIONS) -> list[str]:
    if len(questions) != expected_count:
        raise ValueError(f'Expected {expected_count} questions, but received {len(questions)}.')
    cleaned_questions = []
    for question_number, question in enumerate(questions, start=1):
        if not isinstance(question, str):
            raise ValueError(f'Question {question_number} is not a string.')
        question = re.sub('\\s+', ' ', question).strip()
        question = re.sub('^\\s*\\d+\\s*[\\.\\)\\-:]\\s*', '', question)
        if not question:
            raise ValueError(f'Question {question_number} is empty.')
        if not question.endswith('?'):
            question += '?'
        cleaned_questions.append(question)
    normalised_questions = [re.sub('[^a-z0-9]+', ' ', question.lower()).strip() for question in cleaned_questions]
    if len(set(normalised_questions)) != expected_count:
        raise ValueError('The response contains duplicate questions.')
    return cleaned_questions
````

````python
def generate_questions(prompt: str, max_retries: int=MAX_RETRIES) -> list[str]:
    latest_error = None
    current_prompt = prompt
    for attempt in range(1, max_retries + 1):
        try:
            result = client.responses.create(model=MODEL_NAME, input=current_prompt)
            response_text = result.output_text
            if not response_text or not response_text.strip():
                raise ValueError('The model returned an empty response.')
            questions = extract_json_array(response_text)
            questions = validate_questions(questions)
            return questions
        except Exception as error:
            latest_error = error
            print(f'Attempt {attempt}/{max_retries} failed: {error}')
            if attempt < max_retries:
                current_prompt = f'\n{prompt}\n\nIMPORTANT CORRECTION:\nYour previous response was invalid.\n\nReturn only one valid JSON array containing exactly\n{NUMBER_OF_QUESTIONS} unique question strings. Do not include headings,\nexplanations, numbering outside the strings, or Markdown code fences.\n'.strip()
                wait_time = attempt * 5
                print(f'Retrying in {wait_time} seconds...')
                time.sleep(wait_time)
    raise RuntimeError(f'Question generation failed after {max_retries} attempts. Last error: {latest_error}')
````

````python
def safe_filename(name: str) -> str:
    name = re.sub('[<>:"/\\\\|?*]', '_', str(name))
    name = re.sub('\\s+', ' ', name).strip()
    name = name.rstrip('. ')
    return name
````

````python
def save_questions(company_name: str, questions: list[str], output_directory: Path) -> Path:
    output_filename = f'{safe_filename(company_name)}.txt'
    output_path = output_directory / output_filename
    output_text = '\n'.join((f'{number}. {question}' for number, question in enumerate(questions, start=1)))
    output_path.write_text(output_text, encoding='utf-8')
    return output_path
````

````python
processing_results = []
for row_number, row in tqdm(companies.iterrows(), total=len(companies), desc='Generating company questions'):
    ticker = str(row['Symbol']).strip()
    company_name = str(row['Security']).strip()
    sector = str(row['GICS Sector']).strip() if pd.notna(row['GICS Sector']) else 'Not provided'
    sub_industry = str(row['GICS Sub-Industry']).strip() if pd.notna(row['GICS Sub-Industry']) else 'Not provided'
    output_path = QUESTIONS_DIR / f'{safe_filename(company_name)}.txt'
    if output_path.exists() and (not OVERWRITE_EXISTING):
        print(f'Skipping {ticker} - output already exists: {output_path.name}')
        processing_results.append({'Ticker': ticker, 'Company': company_name, 'Status': 'Skipped - output exists', 'Filing': '', 'Output': str(output_path), 'Error': ''})
        continue
    filing_path = find_company_filing(ticker, filing_index)
    if filing_path is None:
        error_message = f'No matching filing found for ticker {ticker}'
        print(error_message)
        processing_results.append({'Ticker': ticker, 'Company': company_name, 'Status': 'Failed', 'Filing': '', 'Output': '', 'Error': error_message})
        continue
    try:
        print(f'\nProcessing {company_name} ({ticker})')
        print(f'Using filing: {filing_path.name}')
        full_filing_text = clean_filing_html(filing_path)
        if len(full_filing_text) < 1000:
            raise ValueError('Very little text was extracted from the filing.')
        selected_filing_text = select_relevant_filing_text(full_filing_text, max_characters=MAX_FILING_CHARACTERS)
        filing_date = extract_filing_date(filing_path)
        prompt = create_question_generation_prompt(company_name=company_name, ticker=ticker, sector=sector, sub_industry=sub_industry, filing_date=filing_date, filing_text=selected_filing_text)
        questions = generate_questions(prompt)
        saved_path = save_questions(company_name=company_name, questions=questions, output_directory=QUESTIONS_DIR)
        print(f'Saved {len(questions)} questions to {saved_path}')
        processing_results.append({'Ticker': ticker, 'Company': company_name, 'Status': 'Completed', 'Filing': filing_path.name, 'Output': str(saved_path), 'Error': ''})
    except Exception as error:
        print(f'Failed to process {company_name} ({ticker}): {error}')
        processing_results.append({'Ticker': ticker, 'Company': company_name, 'Status': 'Failed', 'Filing': filing_path.name, 'Output': '', 'Error': str(error)})
    time.sleep(DELAY_BETWEEN_REQUESTS)
````

````python
results_df = pd.DataFrame(processing_results)
report_path = QUESTIONS_DIR / 'generation_report.csv'
results_df.to_csv(report_path, index=False)
print('\nProcessing summary:')
print(results_df['Status'].value_counts(dropna=False))
print(f'\nReport saved to: {report_path}')
display(results_df)
````
